```python
def get_embedding(text):
    """Get embedding from OpenAI API"""
    response = openai_client.embeddings.create(
        model="text-embedding-3-small",
        input=text.replace("\n", " ")
    )
    return response.data[0].embedding

# Generate an embedding for a sample chunk
sample_embedding = get_embedding(chunks[0])
print(f"Generated embedding with {len(sample_embedding)} dimensions")
print(f"length==={len(sample_embedding)}")
print(f"First few values of embedding: {sample_embedding[:5]}")
```

```python
def create_vectors_from_pdf(chunks, source_path, max_chunks=None):
    """Create vectors in Pinecone from text chunks"""
    if max_chunks:
        chunks = chunks[:max_chunks]
        print(f"Processing first {max_chunks} chunks only")

    vector_ids = []
    for i, chunk in enumerate(chunks):
        # Generate ID
        vector_id = str(uuid.uuid4())
        vector_ids.append(vector_id)

        # Get embedding
        embedding = get_embedding(chunk)

        # Upsert to Pinecone
        index.upsert(
            vectors=[{
                "id": vector_id,
                "values": embedding,
                "metadata": {
                    "text": chunk,
                    "source": source_path,
                    "chunk_index": i
                }
            }]
        )
        print(f"Uploaded vector {i+1}/{len(chunks)} with ID: {vector_id}")

    print(f"Created {len(vector_ids)} vectors in Pinecone")
    return vector_ids

# Create vectors from first 3 chunks (for demonstration)
print("\n=== CREATE OPERATION ===")
vector_ids = create_vectors_from_pdf(chunks, pdf_path, max_chunks=4)
print(f"Vector IDs created: {vector_ids}")
```

```python
# Step 5: READ Operation - Search for similar content
def search_similar(query, top_k=3):
    """Search for similar content using a query"""
    print(f"Searching for: '{query}'")

    # Get query embedding
    query_embedding = get_embedding(query)

    # Search index
    results = index.query(
        vector=query_embedding,
        top_k=top_k,
        include_metadata=True
    )

    # Format results
    formatted_results = []
    print(f"Found {len(results.matches)} matches:")

    for i, match in enumerate(results.matches):
        result = {
            "id": match.id,
            "score": match.score,
            "text": match.metadata.get("text", ""),
            "chunk_index": match.metadata.get("chunk_index", -1)
        }
        formatted_results.append(result)

        print(f"\nMatch {i+1} (Score: {match.score:.4f}):")
        print(f"ID: {match.id}")
        print(f"Chunk Index: {match.metadata.get('chunk_index', -1)}")
        # print(f"Text Preview: {match.metadata.get('text', '')[:150]}...")
        print(f"whole chunk: {match.metadata.get('text', '')}...")

    return formatted_results

# Search for similar content
print("\n=== READ OPERATION ===")
query = "skills and experience"
search_results = search_similar(query)
```

```python
# Step 6: UPDATE Operation - Update vector with new text
def update_vector(vector_id, new_text):
    """Update a vector with new text"""
    print(f"Updating vector with ID: {vector_id}")

    print(f"New text : {new_text}")

    # Get new embedding
    new_embedding = get_embedding(new_text)

    # Fetch existing vector
    fetch_response = index.fetch(ids=[vector_id])

    if vector_id not in fetch_response.vectors:
        print(f"Vector {vector_id} not found")
        return False

    # Get existing metadata
    metadata = fetch_response.vectors[vector_id].metadata
    original_text = metadata.get("text", "")
    print(f"Original text : {original_text}")

    # Create new metadata with updated text
    updated_metadata = dict(metadata)
    updated_metadata["text"] = new_text
    updated_metadata["updated"] = True

    # Upsert updated vector
    index.upsert(
        vectors=[{
            "id": vector_id,
            "values": new_embedding,
            "metadata": updated_metadata
        }]
    )

    print(f"Successfully updated vector {vector_id}")
    return True

# Update the first search result
print("\n=== UPDATE OPERATION ===")
if search_results:
    vector_to_update = search_results[2]["id"]
    print("je vector update korbo seitar vector_id=",vector_to_update)
    original_text = search_results[0]["text"]
    # new_text = original_text + " [UPDATED]"
    new_text="This is alamin who is a machine learning engineer of JBC"
    update_success = update_vector(vector_to_update, new_text)
    query = "who is alamin"
    # Verify update
    if update_success:
        print("\nVerifying update:")
        verification = search_similar(query)
        for result in verification:
            if result["id"] == vector_to_update and "[UPDATED]" in result["text"]:
                print("✅ Update verified successfully!")
```

```python
# Step 7: DELETE Operation - Delete vectors
def delete_vectors(vector_ids):
    """Delete vectors from index"""
    print(f"Deleting {len(vector_ids)} vectors: {vector_ids}")

    # Delete vectors
    index.delete(ids=vector_ids)

    # Verify deletion
    fetch_response = index.fetch(ids=vector_ids)
    remaining = len(fetch_response.vectors)

    if remaining == 0:
        print(f"✅ Successfully deleted all {len(vector_ids)} vectors")
    else:
        print(f"⚠️ {remaining} out of {len(vector_ids)} vectors still exist")

    return remaining == 0

# Delete one vector
print("\n=== DELETE OPERATION ===")
if vector_ids:
    # Delete the first vector
    print("first_vector=={vector_ids[0]}")
    vector_to_delete = vector_ids[0]
    delete_success = delete_vectors([vector_to_delete])

    # Remove from our list if successful
    if delete_success:
        vector_ids.remove(vector_to_delete)
        print(f"Remaining vector IDs: {vector_ids}")
```

```python
# Simplified implementation of encrypted embeddings with Pinecone
import openai
from pinecone import Pinecone
import numpy as np
from cryptography.fernet import Fernet



# PINECONE_ENVIRONMENT = "gcp-starter"  # e.g., "gcp-starter"


# Initialize API clients
openai_client = openai.OpenAI(api_key="")
pc = Pinecone(api_key="")

# Set up encryption
# IMPORTANT: Store this key securely (e.g., in environment variables)
ENCRYPTION_KEY = Fernet.generate_key()  # Generate a key or use a fixed key
cipher = Fernet(ENCRYPTION_KEY)

# Create or connect to Pinecone index
index = pc.Index("testing")
```

```python


# 1. Function to encrypt embeddings
def encrypt_embedding(embedding):
    """Encrypt an embedding while preserving similarity relationships"""
    # Convert embedding to numpy array
    embedding_array = np.array(embedding)

    # Generate a random orthogonal transformation matrix (preserves distances and angles)
    random_matrix = np.random.normal(0, 1, (len(embedding), len(embedding)))
    q, _ = np.linalg.qr(random_matrix)  # QR decomposition gives orthogonal matrix

    # Apply transformation
    transformed = np.dot(q, embedding_array)

    print(f"transformed list===={transformed.tolist()}")

    # Return transformed vector
    return transformed.tolist()

# 2. Function to get embedding from OpenAI
def get_embedding(text):
    """Get embedding from OpenAI API"""
    response = openai_client.embeddings.create(
        model="text-embedding-3-small",
        input=text
    )
    return response.data[0].embedding

# 3. Function to process chunks with encryption
def process_and_store_encrypted(text_chunks):
    """Process text chunks, encrypt embeddings, and store in Pinecone"""
    vector_ids = []

    for i, chunk in enumerate(text_chunks):
        # Get original embedding
        original_embedding = get_embedding(chunk)

        # Encrypt the embedding
        encrypted_embedding = encrypt_embedding(original_embedding)

        # Store in Pinecone
        vector_id = f"chunk-{i}"
        index.upsert(
            vectors=[{
                "id": vector_id,
                "values": encrypted_embedding,
                "metadata": {
                    "encrypted": True,
                    "chunk_index": i
                    # Don't store original text for security
                }
            }]
        )

        vector_ids.append(vector_id)
        print(f"Processed chunk {i+1}")

    return vector_ids

# 4. Function to search with encrypted query
def search_encrypted(query_text, top_k=3):
    """Search using an encrypted query embedding"""
    # Get query embedding
    query_embedding = get_embedding(query_text)
    print(f"query_embedding==={query_embedding}\n")

    # Encrypt query using same method
    encrypted_query = encrypt_embedding(query_embedding)
    print(f"encrypted_query==={query_embedding}\n")

    # Search with encrypted query
    results = index.query(
        vector=encrypted_query,
        top_k=top_k,
        include_metadata=True
    )
    print(f"result==={results}\n")
    return results

# Example usage
sample_chunks = [
    "Machine learning models can process large datasets",
    "Neural networks have multiple layers of neurons",
    "Privacy-preserving machine learning protects user data",
    "Vector databases enable efficient similarity search"
]

# Store encrypted embeddings
vector_ids = process_and_store_encrypted(sample_chunks)

# Search with encrypted query
search_results = search_encrypted("How do neural networks work?", top_k=2)

# Display results
print("\nSearch results:")
for i, match in enumerate(search_results.matches):
    print(f"Match {i+1} (Score: {match.score:.4f})")
    print(f"ID: {match.id}")
    print(f"Metadata: {match.metadata}")
```

```python
# Encrypted embeddings with text decryption for search results
import openai
from pinecone import Pinecone
import numpy as np
from cryptography.fernet import Fernet
import json
import base64




# Initialize API clients
openai_client = openai.OpenAI(api_key="")
pc = Pinecone(api_key="")

# Set up encryption
# IMPORTANT: Store this key securely (e.g., in environment variables)
ENCRYPTION_KEY = Fernet.generate_key()  # Generate a key or use a fixed key
cipher = Fernet(ENCRYPTION_KEY)

# Create or connect to Pinecone index
index = pc.Index("testing")

# Function to encrypt text
def encrypt_text(text):
    """Encrypt text to store securely"""
    # Convert text to bytes and encrypt
    encrypted_bytes = cipher.encrypt(text.encode('utf-8'))
    # Convert to base64 string for storage
    return base64.b64encode(encrypted_bytes).decode('utf-8')

# Function to decrypt text
def decrypt_text(encrypted_text):
    """Decrypt text from encrypted storage"""
    try:
        # Convert from base64 string to bytes
        encrypted_bytes = base64.b64decode(encrypted_text.encode('utf-8'))
        # Decrypt bytes and convert to string
        decrypted_bytes = cipher.decrypt(encrypted_bytes)
        return decrypted_bytes.decode('utf-8')
    except Exception as e:
        print(f"Error decrypting text: {e}")
        return "**Decryption failed**"

# Function to encrypt embeddings (same as before)
def encrypt_embedding(embedding):
    """Encrypt an embedding while preserving similarity relationships"""
    # Convert embedding to numpy array
    embedding_array = np.array(embedding)
    print(f"embedding_array===={embedding_array}\n")

    # Generate a random orthogonal transformation matrix (preserves distances and angles)
    random_matrix = np.random.normal(0, 1, (len(embedding), len(embedding)))
    q, _ = np.linalg.qr(random_matrix)  # QR decomposition gives orthogonal matrix

    # Apply transformation
    transformed = np.dot(q, embedding_array)
    print(f"transformed===={transformed}\n")
    print(f"transformed_to_list===={transformed.tolist()}\n")
    # Return transformed vector
    return transformed.tolist()

# Function to get embedding from OpenAI
def get_embedding(text):
    """Get embedding from OpenAI API"""
    response = openai_client.embeddings.create(
        model="text-embedding-3-small",
        input=text
    )
    return response.data[0].embedding

# Process and store encrypted chunks with encrypted text
def process_and_store_encrypted(text_chunks, storage_method="encrypted_metadata"):
    """
    Process text chunks, encrypt embeddings, and store in Pinecone

    Args:
        text_chunks: List of text chunks to process
        storage_method: How to store original text
            - "encrypted_metadata": Store encrypted text in metadata (retrievable)
            - "separate_storage": Store mapping in separate secure storage (not implemented here)
            - "no_storage": Don't store original text (most secure but not retrievable)
    """
    vector_ids = []

    for i, chunk in enumerate(text_chunks):
        # Get original embedding
        original_embedding = get_embedding(chunk)

        # Encrypt the embedding
        encrypted_embedding = encrypt_embedding(original_embedding)
        print(f"encrypted_embedding===={encrypted_embedding}\n")

        # Prepare metadata
        metadata = {
            "encrypted": True,
            "chunk_index": i
        }

        # Handle original text based on storage method
        if storage_method == "encrypted_metadata":
            # Encrypt the text and store in metadata

            metadata["encrypted_text"] = encrypt_text(chunk)
            print(f"chunk===={chunk}\n")
            print(f"encrypt_text===={encrypt_text(chunk)}\n")
        elif storage_method == "separate_storage":
            # In a real implementation, you would store this in a separate secure database
            # with vector_id as the key
            pass
        # For "no_storage", we don't store the text at all

        # Store in Pinecone
        vector_id = f"chunk-{i}"
        index.upsert(
            vectors=[{
                "id": vector_id,
                "values": encrypted_embedding,
                "metadata": metadata
            }]
        )

        vector_ids.append(vector_id)
        print(f"Processed chunk {i+1}")

    return vector_ids

# Search with encrypted query and decrypt results
def search_encrypted_with_decryption(query_text, top_k=3):
    """Search using an encrypted query embedding and decrypt results"""
    # Get query embedding
    query_embedding = get_embedding(query_text)
    print(f"query_embedding===={query_embedding}\n")

    # Encrypt query using same method
    encrypted_query = encrypt_embedding(query_embedding)
    print(f"encrypted_query===={encrypted_query}\n")

    # Search with encrypted query
    results = index.query(
        vector=encrypted_query,
        top_k=top_k,
        include_metadata=True
    )

    # Process results to include decrypted text where available
    decrypted_results = []

    for i, match in enumerate(results.matches):
        result = {
            "id": match.id,
            "score": match.score,
            "metadata": match.metadata
        }

        # Check if this is an encrypted result with stored encrypted text
        if match.metadata.get("encrypted") and "encrypted_text" in match.metadata:
            # Decrypt the text
            encrypted_text = match.metadata["encrypted_text"]
            decrypted_text = decrypt_text(encrypted_text)
            result["decrypted_text"] = decrypted_text

        decrypted_results.append(result)

        # Print result info
        print(f"Match {i+1} (Score: {match.score:.4f})")
        print(f"ID: {match.id}")

        if "decrypted_text" in result:
            print(f"Decrypted Text: {result['decrypted_text'][:150]}...")
        else:
            print("No decrypted text available")

        print()

    return decrypted_results

# Example usage
sample_chunks = [
    "Machine learning models can process large datasets",
    "Neural networks have multiple layers of neurons",
    "Privacy-preserving machine learning protects user data",
    "Vector databases enable efficient similarity search"
]

# Store encrypted embeddings with encrypted text
print("Processing and storing encrypted chunks...")
vector_ids = process_and_store_encrypted(sample_chunks, storage_method="encrypted_metadata")

# Search with encrypted query and decrypt results
print("\nSearching with encrypted query...")
search_results = search_encrypted_with_decryption("How do neural networks work?", top_k=2)
```
